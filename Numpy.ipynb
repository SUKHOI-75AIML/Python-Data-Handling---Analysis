```python
import numpy as np

arr = np.random.randint(1, 21, size=(5, 5))

print("Original Array:")
print(arr)

arr[:, 2] = 1

print("Modified Array:")
print(arr)
```

```python
import numpy as np

arr = np.arange(1, 17).reshape(4, 4)

print("Original Array:")
print(arr)

np.fill_diagonal(arr, 0)

print("Modified Array:")
print(arr)
```

```python
import numpy as np

arr = np.random.randint(1, 50, size=(5, 5))

print("Original Array:")
print(arr)

mask = np.zeros(arr.shape, dtype=bool)

mask[0, :] = True
mask[-1, :] = True
mask[:, 0] = True
mask[:, -1] = True

border = arr[mask]

print("Border Elements:")
print(border)
```

```python
import numpy as np

arr = np.arange(1, 17).reshape(4, 4)

print("Original Array:")
print(arr)

row_sum = np.sum(arr, axis=1)
column_sum = np.sum(arr, axis=0)

print("Row-wise Sum:")
print(row_sum)

print("Column-wise Sum:")
print(column_sum)
```

```python
import numpy as np

# Create a 3x3 array
arr = np.arange(1, 10).reshape(3, 3)

print("Original Array:")
print(arr)

# Reshape into 1 row and 9 columns
arr1 = arr.reshape(1, 9)

print("\nArray of Shape (1, 9):")
print(arr1)

# Reshape into 9 rows and 1 column
arr2 = arr1.reshape(9, 1)

print("\nArray of Shape (9, 1):")
print(arr2)
```
