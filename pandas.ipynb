```python
import pandas as pd
import numpy as np

# Create a DataFrame with random integers
df = pd.DataFrame(
    np.random.randint(1, 10, size=(5, 3)),
    columns=['A', 'B', 'C']
)

print("Original DataFrame:")
print(df)

# Add a new column with the product of A and B
df['Product'] = df['A'] * df['B']

print("\nDataFrame after adding Product column:")
print(df)
```

```python
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame(
    np.random.randint(1, 10, size=(4, 3)),
    columns=['A', 'B', 'C']
)

print("Original DataFrame:")
print(df)

# Calculate row-wise sum
print("\nRow-wise Sum:")
print(df.sum(axis=1))

# Calculate column-wise sum
print("\nColumn-wise Sum:")
print(df.sum(axis=0))
```

```python
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame(
    np.random.randint(1, 20, size=(5, 3)),
    columns=['A', 'B', 'C']
)

# Introduce NaN values
df.loc[1, 'A'] = np.nan
df.loc[2, 'B'] = np.nan
df.loc[3, 'C'] = np.nan

print("Original DataFrame:")
print(df)

# Fill NaN values with column mean
df = df.fillna(df.mean())

print("\nDataFrame after filling NaN:")
print(df)
```

```python
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame(
    np.random.randint(1, 50, size=(6, 4)),
    columns=['A', 'B', 'C', 'D']
)

# Introduce NaN values
df.loc[1, 'A'] = np.nan
df.loc[3, 'C'] = np.nan
df.loc[4, 'B'] = np.nan

print("Original DataFrame:")
print(df)

# Drop rows containing NaN
df = df.dropna()

print("\nDataFrame after dropping NaN rows:")
print(df)
```

```python
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Value': np.random.randint(1, 100, size=10)
})

print("Original DataFrame:")
print(df)

# Group by Category and calculate sum and mean
result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print("\nData Aggregation:")
print(result)
```

```python
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame({
    'Product': ['Laptop', 'Mobile', 'Tablet', 'Laptop',
                'Mobile', 'Tablet'],
    'Category': ['Electronics', 'Electronics', 'Accessories',
                 'Electronics', 'Accessories', 'Accessories'],
    'Sales': np.random.randint(100, 1000, size=6)
})

print("Original DataFrame:")
print(df)

# Group by Category and calculate total sales
result = df.groupby('Category')['Sales'].sum()

print("\nTotal Sales by Category:")
print(result)
```

```python
import pandas as pd

# Create first DataFrame
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Aishwarya', 'Rahul', 'Priya', 'Sneha']
})

# Create second DataFrame
df2 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Marks': [85, 90, 88, 95]
})

print("First DataFrame:")
print(df1)

print("\nSecond DataFrame:")
print(df2)

# Merge DataFrames using common column
merged_df = pd.merge(df1, df2, on='ID')

print("\nMerged DataFrame:")
print(merged_df)
```

```python
import pandas as pd

# Create MultiIndex
index = pd.MultiIndex.from_tuples([
    ('A', 1),
    ('A', 2),
    ('B', 1),
    ('B', 2),
    ('C', 1),
    ('C', 2)
], names=['Category', 'Number'])

# Create DataFrame
df = pd.DataFrame({
    'Value': [10, 20, 30, 40, 50, 60]
}, index=index)

print("Original MultiIndex DataFrame:")
print(df)

# Indexing
print("\nData for Category A:")
print(df.loc['A'])

# Slicing
print("\nData from Category A to B:")
print(df.loc[('A', 1):('B', 2)])
```

```python
import pandas as pd
import numpy as np

# Create MultiIndex
index = pd.MultiIndex.from_tuples([
    ('A', 'A1'),
    ('A', 'A2'),
    ('A', 'A1'),
    ('B', 'B1'),
    ('B', 'B2'),
    ('B', 'B1'),
    ('C', 'C1'),
    ('C', 'C2')
], names=['Category', 'SubCategory'])

# Create DataFrame
df = pd.DataFrame({
    'Value': np.random.randint(10, 100, size=8)
}, index=index)

print("Original DataFrame:")
print(df)

# Calculate sum for each Category and SubCategory
result = df.groupby(level=['Category', 'SubCategory']).sum()

print("\nSum of Values:")
print(result)
```

```python
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame(
    np.random.randint(1, 20, size=(5, 3)),
    columns=['A', 'B', 'C']
)

print("Original DataFrame:")
print(df)

# Apply function to double values
result = df.apply(lambda x: x * 2)

print("\nDataFrame after doubling values:")
print(result)
```

```python
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame(
    np.random.randint(1, 20, size=(6, 3)),
    columns=['A', 'B', 'C']
)

print("Original DataFrame:")
print(df)

# Create a new column using lambda function
df['Total'] = df.apply(
    lambda row: row['A'] + row['B'] + row['C'],
    axis=1
)

print("\nDataFrame after adding Total column:")
print(df)
```
